# 🟢 Easy: Cross-Entropy Loss

Implement **cross-entropy loss** from scratch.

$$\text{CE}(x, y) = -\log\frac{e^{x_y}}{\sum_j e^{x_j}}$$

---

## Notes

### 1. Logits version

This formula is for:

- logits $x$
- correct class index $y$

It **includes softmax** inside the equation, since:

$$
\text{softmax}(x)_y = \frac{e^{x_y}}{\sum_j e^{x_j}}
$$

So equivalently:

$$
\text{CE}(x, y) = -\log(\text{softmax}(x)_y)
$$

This is also the **negative log-likelihood (NLL)** of the correct class.

---

### 2. Probability version

If the goal is to align the predicted probability distribution $\hat{y}$ with the target probability distribution $y$(over i classes), we use:

$$
\text{CE}(y, \hat{y}) = -\sum_i y_i \log(\hat{y}_i)
$$

This version **does not include softmax**, because $\hat{y}$ is already a probability distribution(after softmax).

If $y$ is one-hot, cross-entropy reduces to:

$$
\text{CE} = -\log(\hat{y}_{\text{correct class}})
$$

So with one-hot labels:

$$
\text{CE} = \text{NLL}
$$

```python
def cross_entropy_loss(logits: Tensor, targets: Tensor) -> Tensor:
    # logits: (B, C) float, targets: (B,) long indices
    # Returns: scalar loss (mean over batch)
```

### Rules
- Do NOT use `F.cross_entropy` or `nn.CrossEntropyLoss`
- Must be numerically stable (use logsumexp trick)

````python
# ✏️ YOUR IMPLEMENTATION HERE

def cross_entropy_loss(logits, targets):
    correct_logits = logits[torch.arange(len(targets)), targets].unsqueeze(-1)
    return torch.mean(- (correct_logits - torch.logsumexp(logits, dim=-1, keepdim=True))) #-( log(exp(x_y)) - log(sum_j exp(x_j)) )
````

````python
# 🧪 Debug
logits = torch.randn(4, 10) # 4 samples, 10 classes, single-label multi-class classification
targets = torch.randint(0, 10, (4,)) 
print('Loss:', cross_entropy_loss(logits, targets))
print('Ref: ', torch.nn.functional.cross_entropy(logits, targets))
````
